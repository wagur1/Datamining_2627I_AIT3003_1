## Problem P1: Euclidean, Manhattan, and Minkowski Distances

> **Prompt:** Cho hai đối tượng được biểu diễn bằng các bộ số $(22, 1, 42, 10)$ và $(20, 0, 36, 8)$:
> - (a) Tính khoảng cách Euclidean giữa hai đối tượng này.
> - (b) Tính khoảng cách Manhattan giữa hai đối tượng này.
> - (c) Tính khoảng cách Minkowski giữa hai đối tượng này, sử dụng $q = 3$.

### Mathematical Formulation
For two $p$-dimensional numeric objects $\mathbf{x}_1 = (x_{11}, x_{12}, \dots, x_{1p})$ and $\mathbf{x}_2 = (x_{21}, x_{22}, \dots, x_{2p})$, the **Minkowski distance** of order $q \ge 1$ is:
$$d_q(\mathbf{x}_1, \mathbf{x}_2) = \left( \sum_{k=1}^{p} |x_{1k} - x_{2k}|^q \right)^{\frac{1}{q}}$$
- **Manhattan ($L_1$) distance ($q = 1$):** $d_1(\mathbf{x}_1, \mathbf{x}_2) = \sum_{k=1}^{p} |x_{1k} - x_{2k}| = 2 + 1 + 6 + 2 = 11$
- **Euclidean ($L_2$) distance ($q = 2$):** $d_2(\mathbf{x}_1, \mathbf{x}_2) = \sqrt{2^2 + 1^2 + 6^2 + 2^2} = \sqrt{45} = 3\sqrt{5} \approx 6.7082$
- **Minkowski ($L_3$) distance ($q = 3$):** $d_3(\mathbf{x}_1, \mathbf{x}_2) = \sqrt[3]{2^3 + 1^3 + 6^3 + 2^3} = \sqrt[3]{233} \approx 6.1534$


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def minkowski_distance(x1: np.ndarray, x2: np.ndarray, q: float) -> float:
    if np.isinf(q):
        return float(np.max(np.abs(x1 - x2)))
    return float(np.sum(np.abs(x1 - x2) ** q) ** (1.0 / q))

x1 = np.array([22, 1, 42, 10], dtype=float)
x2 = np.array([20, 0, 36, 8], dtype=float)
diffs = np.abs(x1 - x2)

d_euclidean = minkowski_distance(x1, x2, q=2)
d_manhattan = minkowski_distance(x1, x2, q=1)
d_minkowski_3 = minkowski_distance(x1, x2, q=3)
d_supremum = minkowski_distance(x1, x2, q=np.inf)

print(f"Object 1: x1 = {tuple(x1.astype(int).tolist())}")
print(f"Object 2: x2 = {tuple(x2.astype(int).tolist())}")
print(f"Absolute differences |x1 - x2| = {tuple(diffs.astype(int).tolist())}")
print("-" * 60)
print(f"(a) Euclidean Distance (q=2) : sqrt({int(np.sum(diffs**2))}) = {d_euclidean:.4f}")
print(f"(b) Manhattan Distance (q=1) : {int(d_manhattan)}")
print(f"(c) Minkowski Distance (q=3) : ({int(np.sum(diffs**3))})^(1/3) = {d_minkowski_3:.4f}")
print(f"[*] Supremum Distance (q=inf): {int(d_supremum)}")
```

```python
# Visualizing Component Contributions and Minkowski Distance Decay
fig, axes = plt.subplots(1, 2, figsize=(12, 4.6), dpi=120)

dims = [f"Dim {i+1}\n({int(x1[i])} vs {int(x2[i])})" for i in range(len(x1))]
x_pos = np.arange(len(dims))
width = 0.26

axes[0].bar(x_pos - width, diffs, width, label=r"$|x_{1k} - x_{2k}|^1$ ($q=1$)", color="#1f77b4", edgecolor="black")
axes[0].bar(x_pos, diffs**2, width, label=r"$|x_{1k} - x_{2k}|^2$ ($q=2$)", color="#ff7f0e", edgecolor="black")
axes[0].bar(x_pos + width, diffs**3, width, label=r"$|x_{1k} - x_{2k}|^3$ ($q=3$)", color="#2ca02c", edgecolor="black")
axes[0].set_yscale("symlog", linthresh=1)
axes[0].set_xticks(x_pos)
axes[0].set_xticklabels(dims)
axes[0].set_ylabel("Power of Absolute Difference (Log Scale)")
axes[0].set_title("P1: Component Contributions by Order $q$", fontweight="bold")
axes[0].grid(True, linestyle=":", alpha=0.6)
axes[0].legend()

q_vals = np.linspace(1.0, 10.0, 200)
d_vals = [minkowski_distance(x1, x2, q) for q in q_vals]
axes[1].plot(q_vals, d_vals, color="#1f77b4", linewidth=2.5, label=r"Minkowski $L_q(x_1, x_2)$")
axes[1].axhline(d_supremum, color="gray", linestyle="--", linewidth=1.5, label=rf"Supremum Limit $L_\infty = {int(d_supremum)}$")

for q_pt, d_pt, lbl, col in [(1, d_manhattan, "Manhattan (q=1)", "#d62728"),
                             (2, d_euclidean, "Euclidean (q=2)", "#ff7f0e"),
                             (3, d_minkowski_3, "Minkowski (q=3)", "#2ca02c")]:
    axes[1].scatter([q_pt], [d_pt], color=col, s=65, zorder=5, edgecolors="black", label=f"{lbl}: {d_pt:.4f}")

axes[1].set_xlabel("Minkowski Order $q$")
axes[1].set_ylabel("Distance $d(x_1, x_2)$")
axes[1].set_title(r"P1: Minkowski Distance Decay as $q \to \infty$", fontweight="bold")
axes[1].grid(True, linestyle=":", alpha=0.6)
axes[1].legend()
plt.tight_layout()
plt.show()
```

## Problem P2: Jaccard Similarity Between Word Sets

> **Prompt:** Cho hai câu sau, hãy tính độ tương tự Jaccard giữa chúng bằng cách coi mỗi câu là một tập từ.
> - (a) “ông già đi nhanh quá”
> - (b) “thời_gian trôi đi nhanh quá”.

### Mathematical Formulation
$$J(A, B) = \frac{|A \cap B|}{|A \cup B|} = \frac{|A \cap B|}{|A| + |B| - |A \cap B|}$$
- **Primary Tokenization (whitespace split, where `thời_gian` is pre-segmented with `_`):**
  - $A = \{\text{``ông''}, \text{``già''}, \text{``đi''}, \text{``nhanh''}, \text{``quá''}\}$ ($|A| = 5$)
  - $B = \{\text{``thời\_gian''}, \text{``trôi''}, \text{``đi''}, \text{``nhanh''}, \text{``quá''}\}$ ($|B| = 5$)
  - $A \cap B = \{\text{``đi''}, \text{``nhanh''}, \text{``quá''}\}$ ($|A \cap B| = 3$)
  - $A \cup B = \{\text{``ông''}, \text{``già''}, \text{``thời\_gian''}, \text{``trôi''}, \text{``đi''}, \text{``nhanh''}, \text{``quá''}\}$ ($|A \cup B| = 7$)
  - $J(A, B) = \frac{3}{7} \approx 0.4286$
- **Alternative Note (if `ông già` is also compound-segmented as `ông_già`):** $|A' \cap B| = 3, |A' \cup B| = 6 \implies J(A', B) = \frac{3}{6} = 0.5000$.


```python
sent_a = "ông già đi nhanh quá"
sent_b = "thời_gian trôi đi nhanh quá"

tokens_a = sent_a.split()
tokens_b = sent_b.split()

set_a, set_b = set(tokens_a), set(tokens_b)
intersection = sorted(set_a & set_b, key=lambda w: (tokens_a + tokens_b).index(w))
union = sorted(set_a | set_b, key=lambda w: (tokens_a + tokens_b).index(w))

jaccard_sim = len(intersection) / len(union)
jaccard_dist = 1.0 - jaccard_sim

print(f"Set A (|A|={len(set_a)}): {tokens_a}")
print(f"Set B (|B|={len(set_b)}): {tokens_b}")
print(f"Intersection A ∩ B (|A ∩ B|={len(intersection)}): {intersection}")
print(f"Union A ∪ B (|A ∪ B|={len(union)}): {union}")
print(f"Jaccard Similarity J(A, B) = {len(intersection)}/{len(union)} ≈ {jaccard_sim:.4f}")
print(f"Jaccard Distance d_J(A, B) = 1 - J(A, B) = {len(union)-len(intersection)}/{len(union)} ≈ {jaccard_dist:.4f}")
```

## Problem P3: Term-Frequency Vectors and Cosine Similarity

> **Prompt:** Biểu diễn các câu sau thành vector tần suất từ và tính độ tương tự cosine giữa chúng:
> - (a) “The sly fox jumped over the lazy dog.”
> - (b) “The dog jumped at the intruder.”
> 
> *Lưu ý: Cần xác định tập từ vựng của cả hai câu. Vector biểu diễn mỗi câu sẽ có số chiều là kích thước tập từ vựng đó (mỗi chiều ứng với một từ, giá trị là tần suất từ trong câu – có thể bằng 0).*

### Mathematical Formulation
- **Vocabulary $V$ (9 words, case-insensitive):** `["the", "sly", "fox", "jumped", "over", "lazy", "dog", "at", "intruder"]`
- **Term-Frequency Vectors:**
  - $\mathbf{d}_a = (2, 1, 1, 1, 1, 1, 1, 0, 0), \quad \|\mathbf{d}_a\| = \sqrt{10} \approx 3.1623$
  - $\mathbf{d}_b = (2, 0, 0, 1, 0, 0, 1, 1, 1), \quad \|\mathbf{d}_b\| = \sqrt{8} = 2\sqrt{2} \approx 2.8284$
- **Dot Product:** $\mathbf{d}_a \cdot \mathbf{d}_b = 2(2) + 1(1) + 1(1) = 6$
- **Cosine Similarity:**
  $$\cos(\mathbf{d}_a, \mathbf{d}_b) = \frac{\mathbf{d}_a \cdot \mathbf{d}_b}{\|\mathbf{d}_a\| \|\mathbf{d}_b\|} = \frac{6}{\sqrt{10}\sqrt{8}} = \frac{6}{\sqrt{80}} = \frac{3\sqrt{5}}{10} \approx 0.6708$$


```python
import re

doc_a = "The sly fox jumped over the lazy dog."
doc_b = "The dog jumped at the intruder."

tokens_a = re.sub(r"[^\w\s]", "", doc_a.lower()).split()
tokens_b = re.sub(r"[^\w\s]", "", doc_b.lower()).split()

vocab = []
for w in tokens_a + tokens_b:
    if w not in vocab:
        vocab.append(w)

tf_a = np.array([tokens_a.count(w) for w in vocab])
tf_b = np.array([tokens_b.count(w) for w in vocab])

df_tf = pd.DataFrame({
    "Word": vocab,
    "TF_a (d_a)": tf_a,
    "TF_b (d_b)": tf_b,
    "d_a * d_b": tf_a * tf_b,
    "d_a^2": tf_a ** 2,
    "d_b^2": tf_b ** 2,
})
print(df_tf.to_string(index=False))

dot_prod = int(np.dot(tf_a, tf_b))
norm_a = float(np.linalg.norm(tf_a))
norm_b = float(np.linalg.norm(tf_b))
cos_sim = dot_prod / (norm_a * norm_b)

print("-" * 60)
print(f"d_a = {tuple(tf_a.tolist())}, ||d_a|| = sqrt({int(np.sum(tf_a**2))}) ≈ {norm_a:.4f}")
print(f"d_b = {tuple(tf_b.tolist())}, ||d_b|| = sqrt({int(np.sum(tf_b**2))}) ≈ {norm_b:.4f}")
print(f"Dot product d_a · d_b = {dot_prod}")
print(f"Cosine Similarity = {dot_prod} / sqrt(80) ≈ {cos_sim:.4f}")
```

## Problem P4: Levenshtein Edit Distance Dynamic Programming Table

> **Prompt:** Lập bảng tính khoảng cách soạn thảo Levenshtein sử dụng ba phép toán: xóa, chèn và thay thế một ký tự trong chuỗi cho hai chuỗi sau “học sinh” và “hojc xih” (Lưu ý: ‘ọ’ cũng tính là một ký tự).

### Dynamic Programming Recurrence
For source string $s_1 = \text{``học sinh''}$ ($m = 8$) and target string $s_2 = \text{``hojc xih''}$ ($n = 8$):
$$D[i, 0] = i, \quad D[0, j] = j$$
$$D[i, j] = \min \begin{cases} D[i-1, j] + 1 & \text{(Delete)} \\ D[i, j-1] + 1 & \text{(Insert)} \\ D[i-1, j-1] + \mathbb{I}(s_1[i] \ne s_2[j]) & \text{(Match / Substitute)} \end{cases}$$


```python
s1 = "học sinh"
s2 = "hojc xih"

m, n = len(s1), len(s2)
dp = np.zeros((m + 1, n + 1), dtype=int)
for i in range(m + 1):
    dp[i, 0] = i
for j in range(n + 1):
    dp[0, j] = j

for i in range(1, m + 1):
    for j in range(1, n + 1):
        cost = 0 if s1[i - 1] == s2[j - 1] else 1
        dp[i, j] = min(
            dp[i - 1, j] + 1,        # Deletion
            dp[i, j - 1] + 1,        # Insertion
            dp[i - 1, j - 1] + cost  # Match / Substitution
        )

row_labels = ["ε"] + [f"'{c}'" if c != " " else "'_'" for c in s1]
col_labels = ["ε"] + [f"'{c}'" if c != " " else "'_'" for c in s2]
dp_df = pd.DataFrame(dp, index=row_labels, columns=col_labels)
print("Levenshtein Edit Distance DP Table (9x9):")
print(dp_df)
print(f"\nMinimum Levenshtein Edit Distance d_L('{s1}', '{s2}') = {dp[m, n]}")
```

```python
# Backtrace optimal edit alignment and visualize the DP table
path = []
i, j = m, n
while i > 0 or j > 0:
    if i > 0 and j > 0 and s1[i - 1] == s2[j - 1] and dp[i, j] == dp[i - 1, j - 1]:
        path.append((i, j, f"Match '{s1[i-1]}'"))
        i -= 1; j -= 1
    elif j > 0 and dp[i, j] == dp[i, j - 1] + 1:
        path.append((i, j, f"Insert '{s2[j-1]}'"))
        j -= 1
    elif i > 0 and dp[i, j] == dp[i - 1, j] + 1:
        path.append((i, j, f"Delete '{s1[i-1]}'"))
        i -= 1
    else:
        path.append((i, j, f"Sub '{s1[i-1]}' -> '{s2[j-1]}'"))
        i -= 1; j -= 1
path.append((0, 0, "Start"))
path.reverse()

for step_idx, (r, c, op) in enumerate(path):
    print(f"Step {step_idx:>2}: Cell ({r}, {c}) [cost = {dp[r, c]}] -> {op}")

fig, ax = plt.subplots(figsize=(8, 6.5), dpi=120)
im = ax.imshow(dp, cmap="YlGnBu", aspect="equal")
plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04, label="Cumulative Edit Cost")
ax.set_xticks(np.arange(len(col_labels)))
ax.set_yticks(np.arange(len(row_labels)))
ax.set_xticklabels(col_labels, fontweight="bold")
ax.set_yticklabels(row_labels, fontweight="bold")
ax.xaxis.tick_top()

path_coords = {(r, c) for r, c, _ in path}
for r in range(m + 1):
    for c in range(n + 1):
        in_p = (r, c) in path_coords
        ax.text(c, r, str(dp[r, c]), ha="center", va="center",
                color="#d62728" if in_p else ("white" if dp[r, c] > 4 else "black"),
                fontweight="bold" if in_p else "normal",
                bbox=dict(boxstyle="round,pad=0.2", facecolor="#fff3cd", edgecolor="#d62728", lw=1.8) if in_p else None)

ax.set_title(f"P4: Levenshtein Edit Distance DP Table (Distance = {dp[m, n]})", fontweight="bold", pad=20)
plt.tight_layout()
plt.show()
```
