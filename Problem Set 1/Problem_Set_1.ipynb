## Problem 1: Univariate Analysis on Age
Suppose that the data for analysis includes the attribute age:
\{13, 15, 16, 16, 19, 20, 20, 21, 22, 22, 25, 25, 25, 25, 30, 33, 33, 35, 35, 35, 35, 36, 40, 45, 46, 52, 70\}


```
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt

age_data = np.array([13, 15, 16, 16, 19, 20, 20, 21, 22, 22, 25, 25, 25, 25, 30,
                     33, 33, 35, 35, 35, 35, 36, 40, 45, 46, 52, 70])
n = len(age_data)

# a. Mean and Median
mean_age = np.mean(age_data)
median_age = np.median(age_data)
print(f'Mean: {mean_age:.4f} (Sum={np.sum(age_data)}/{n})')
print(f'Median: {median_age}')

# b. Mode and Modality
vals, counts = np.unique(age_data, return_counts=True)
modes = vals[counts == np.max(counts)]
print(f'Modes: {modes} (count={np.max(counts)}), Distribution is Bimodal')

# c. Midrange
midrange_age = (np.min(age_data) + np.max(age_data)) / 2.0
print(f'Midrange: {midrange_age}')

# d. Quartiles (Q1, Q3)
# 27 observations: 14th is median. Lower 13 values: 7th is Q1 (20). Upper 13 values: 21st is Q3 (35).
q1_age = age_data[6]
q3_age = age_data[20]
iqr_age = q3_age - q1_age
print(f'Q1: {q1_age}, Q3: {q3_age}, IQR: {iqr_age}')

# e. Five-number summary
print(f'Five-number summary: [Min={np.min(age_data)}, Q1={q1_age}, Median={median_age}, Q3={q3_age}, Max={np.max(age_data)}]')

# g. Variance and Standard Deviation
print(f'Sample Variance: {np.var(age_data, ddof=1):.4f}, Sample Std: {np.std(age_data, ddof=1):.4f}')
print(f'Population Variance: {np.var(age_data, ddof=0):.4f}, Population Std: {np.std(age_data, ddof=0):.4f}')
```

```
# f. Boxplot of the Data
fig, ax = plt.subplots(figsize=(7, 3), dpi=120)
ax.boxplot(age_data, vert=False, patch_artist=True,
           boxprops=dict(facecolor='#dbeafe', color='#1d4ed8'),
           medianprops=dict(color='#b91c1c', linewidth=2),
           flierprops=dict(marker='o', markerfacecolor='#ef4444', markersize=8))
ax.set_title('Problem 1: Boxplot of Age Attribute', fontsize=12, fontweight='bold')
ax.set_xlabel('Age')
ax.set_yticks([])
ax.grid(axis='x', linestyle='--', alpha=0.5)

# Annotations
ax.text(13, 1.08, 'Min=13', ha='center', fontsize=9, color='#1e3a8a')
ax.text(20, 1.15, 'Q1=20', ha='center', fontsize=9, color='#1e3a8a')
ax.text(25, 1.22, 'Median=25', ha='center', fontsize=9, color='#b91c1c')
ax.text(35, 1.15, 'Q3=35', ha='center', fontsize=9, color='#1e3a8a')
ax.text(52, 1.08, 'Whisker=52', ha='center', fontsize=9, color='#1e3a8a')
ax.text(70, 1.08, 'Outlier=70', ha='center', fontsize=9, color='#ef4444')
plt.tight_layout()
plt.show()
```

## Problem 2: Bivariate Analysis on Age and %Fat
Data of 18 randomly selected adults:


```
age = np.array([23, 23, 27, 27, 39, 41, 47, 49, 50, 52, 54, 54, 56, 57, 58, 58, 60, 61], dtype=float)
fat = np.array([9.5, 26.5, 7.8, 17.8, 31.4, 25.9, 27.4, 27.2, 31.2, 34.6, 42.5, 28.8, 33.4, 30.2, 34.1, 32.9, 41.2, 35.7], dtype=float)

df_p2 = pd.DataFrame({'Age': age, '%Fat': fat})
print('Summary Statistics:')
stats_p2 = pd.DataFrame({
    'Mean': [np.mean(age), np.mean(fat)],
    'Median': [np.median(age), np.median(fat)],
    'Sample Var': [np.var(age, ddof=1), np.var(fat, ddof=1)],
    'Sample Std': [np.std(age, ddof=1), np.std(fat, ddof=1)],
    'Pop Var': [np.var(age, ddof=0), np.var(fat, ddof=0)],
    'Pop Std': [np.std(age, ddof=0), np.std(fat, ddof=0)]
}, index=['Age', '%Fat'])
print(stats_p2.round(4))

# Correlation coefficients
r_pearson, _ = stats.pearsonr(age, fat)
r_spearman, _ = stats.spearmanr(age, fat)
print(f'\nPearson Correlation Coefficient (r): {r_pearson:.4f}')
print(f'Spearman Rank Correlation (rho): {r_spearman:.4f}')
```

```
# Boxplots and Scatter Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4), dpi=120)
ax1.boxplot([age, fat], tick_labels=['Age', '%Fat'], patch_artist=True,
            boxprops=dict(facecolor='#e0e7ff', color='#3730a3'),
            medianprops=dict(color='#dc2626', linewidth=2),
            flierprops=dict(marker='o', markerfacecolor='#ef4444', markersize=7))
ax1.set_title('Boxplots of Age and %Fat', fontweight='bold')
ax1.set_ylabel('Value')
ax1.grid(axis='y', linestyle='--', alpha=0.5)

ax2.scatter(age, fat, color='#2563eb', edgecolors='black', s=50, alpha=0.8)
m, b = np.polyfit(age, fat, 1)
ax2.plot(age, m*age + b, color='#dc2626', linestyle='--', label=f'Trendline (r={r_pearson:.2f})')
ax2.set_title('Scatter Plot: %Fat vs. Age', fontweight='bold')
ax2.set_xlabel('Age')
ax2.set_ylabel('%Fat')
ax2.grid(True, linestyle='--', alpha=0.5)
ax2.legend()
plt.tight_layout()
plt.show()
```

## Problem 3: Cristiano Ronaldo UCL Statistics (03/04 to 21/22)
Dataset of appearances, goals, and assists across 19 Champions League seasons:


```
data_p3 = [
    [5, 0, 1], [7, 0, 2], [6, 0, 0], [11, 3, 5], [11, 8, 1],
    [12, 4, 3], [6, 7, 1], [12, 6, 4], [10, 10, 3], [12, 12, 1],
    [11, 17, 5], [12, 10, 4], [12, 16, 4], [13, 12, 6], [13, 15, 3],
    [9, 6, 2], [8, 4, 1], [6, 4, 2], [7, 6, 0]
]
df_p3 = pd.DataFrame(data_p3, columns=['Appearances', 'Goals', 'Assists'])

print('Summary Statistics:')
for col in df_p3.columns:
    v = df_p3[col].values
    val, cnt = np.unique(v, return_counts=True)
    m = val[cnt == np.max(cnt)]
    sorted_v = np.sort(v)
    print(f'{col}:')
    print(f'  Mean: {np.mean(v):.4f}, Median: {np.median(v)}, Mode: {m}')
    print(f'  Sample Var: {np.var(v, ddof=1):.4f}, Sample Std: {np.std(v, ddof=1):.4f}')
    print(f'  Q1: {sorted_v[4]}, Q3: {sorted_v[14]}')

print('\nSample Covariance Matrix (ddof=1):')
print(df_p3.cov().round(4))
print('\nPopulation Covariance Matrix (ddof=0):')
print(df_p3.cov(ddof=0).round(4))
```

## Problem 4: Pearson's Chi-Square Test of Independence
Contingency table of 800 students (Math, History, CS) vs. Game preferences:


```
obs = np.array([
    [130, 100],  # Math
    [35, 165],   # History
    [280, 90]    # CS
])

row_totals = obs.sum(axis=1)
col_totals = obs.sum(axis=0)
grand_total = obs.sum()

# a. Expected Frequencies
expected = np.outer(row_totals, col_totals) / grand_total
df_exp = pd.DataFrame(expected, index=['Math', 'History', 'CS'], columns=['Like games', 'Do not like games'])
print('Expected Frequencies:')
print(df_exp.round(4))

# b. Chi-Square Test
chi2_stat, p_val, dof, _ = stats.chi2_contingency(obs, correction=False)
crit_05 = stats.chi2.ppf(0.95, df=dof)
crit_01 = stats.chi2.ppf(0.99, df=dof)

print(f'\nChi-Square Statistic: {chi2_stat:.4f}')
print(f'Degrees of Freedom: {dof}')
print(f'p-value: {p_val:.4e}')
print(f'Critical Value (alpha=0.05): {crit_05:.4f}')
print(f'Critical Value (alpha=0.01): {crit_01:.4f}')
print('Conclusion: Since chi2_stat >> crit_05, reject H0. Strong correlation confirmed!')
```
